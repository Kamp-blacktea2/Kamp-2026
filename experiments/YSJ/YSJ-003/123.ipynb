```python
# ============================================================
# AutoEncoder 전체 코드
# Force / Current / Voltage / WELD TIME만 사용
# working time은 절대 사용하지 않음
#
# AutoEncoder
# → 행별 복원오차
# → KMeans(k=3)
# → C0/C1/C2
# → C1/C2에서 Current / Voltage / Weld Time 원인 분석
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.callbacks import EarlyStopping


# ------------------------------------------------------------
# 0. 랜덤 고정
# ------------------------------------------------------------

np.random.seed(42)
tf.random.set_seed(42)


# ------------------------------------------------------------
# 1. Excel 파일 찾기
# ------------------------------------------------------------

possible_paths = [
    Path("Welding Data Set_01.xlsx"),
    Path("../YSJ-002/Welding Data Set_01.xlsx"),
    Path("../Welding Data Set_01.xlsx"),
    Path("../../YSJ-002/Welding Data Set_01.xlsx"),
]

file_path = None

for p in possible_paths:
    if p.exists():
        file_path = p
        break

if file_path is None:
    raise FileNotFoundError(
        "Welding Data Set_01.xlsx 파일을 찾지 못했습니다."
    )

print("=" * 80)
print("사용 파일:", file_path.resolve())
print("=" * 80)


# ------------------------------------------------------------
# 2. Raw data 불러오기
# ------------------------------------------------------------

df = pd.read_excel(
    file_path,
    sheet_name="Raw data"
)

print("\n전체 데이터:", df.shape)

print("\n실제 컬럼명:")
for c in df.columns:
    print(repr(c))


# ------------------------------------------------------------
# 3. 컬럼 자동 탐색
#
# 핵심:
# working time은 제외
# weld + time이 들어간 컬럼만 Weld Time으로 선택
# ------------------------------------------------------------

def normalize_col(col):
    return (
        str(col)
        .lower()
        .replace(" ", "")
        .replace("_", "")
        .replace("-", "")
    )


def find_weld_column(keyword):

    matches = []

    for col in df.columns:

        normalized = normalize_col(col)

        # 반드시 weld가 포함되어야 함
        if "weld" not in normalized:
            continue

        # 원하는 키워드도 포함
        if keyword in normalized:
            matches.append(col)

    if len(matches) == 0:
        return None

    return matches[0]


force_col = find_weld_column("force")
current_col = find_weld_column("current")
voltage_col = find_weld_column("voltage")
weld_time_col = find_weld_column("time")


print("\n" + "=" * 80)
print("AutoEncoder에 실제로 사용할 컬럼")
print("=" * 80)

print("Force     :", force_col)
print("Current   :", current_col)
print("Voltage   :", voltage_col)
print("Weld Time :", weld_time_col)


# ------------------------------------------------------------
# 4. 컬럼 확인
# ------------------------------------------------------------

found_cols = {
    "Force": force_col,
    "Current": current_col,
    "Voltage": voltage_col,
    "Weld_Time": weld_time_col
}

missing = [
    name
    for name, col in found_cols.items()
    if col is None
]

if missing:
    raise ValueError(
        f"다음 변수를 찾지 못했습니다: {missing}\n"
        f"실제 컬럼: {df.columns.tolist()}"
    )


# working time 방어
if "working" in normalize_col(weld_time_col):
    raise ValueError(
        f"잘못된 시간 컬럼을 잡았습니다: {weld_time_col}"
    )


# ------------------------------------------------------------
# 5. 필요한 4개 변수만 가져오기
# ------------------------------------------------------------

data = df[
    [
        force_col,
        current_col,
        voltage_col,
        weld_time_col
    ]
].copy()

data.columns = [
    "Force",
    "Current",
    "Voltage",
    "Weld_Time"
]

# 원래 데이터 행 위치
data["Original_Row"] = np.arange(len(data))


# ------------------------------------------------------------
# 6. 숫자로 변환
# ------------------------------------------------------------

for col in [
    "Force",
    "Current",
    "Voltage",
    "Weld_Time"
]:

    data[col] = pd.to_numeric(
        data[col],
        errors="coerce"
    )


# ------------------------------------------------------------
# 7. 반드시 Weld Time 확인
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("Weld Time 확인")
print("=" * 80)

print(data["Weld_Time"].head(10))

print("\nWeld Time 통계:")
print(data["Weld_Time"].describe())


# ------------------------------------------------------------
# 8. 결측치 확인 / 제거
# ------------------------------------------------------------

features = [
    "Force",
    "Current",
    "Voltage",
    "Weld_Time"
]

print("\n결측치:")
print(data[features].isna().sum())

data = (
    data
    .dropna(subset=features)
    .reset_index(drop=True)
)

print("\n최종 사용 데이터:", len(data))


# ------------------------------------------------------------
# 9. X 생성
# ------------------------------------------------------------

X = data[features].values


# ------------------------------------------------------------
# 10. Train / Test
# ------------------------------------------------------------

indices = np.arange(len(X))

train_idx, test_idx = train_test_split(
    indices,
    test_size=0.20,
    random_state=42
)

X_train = X[train_idx]
X_test = X[test_idx]

print("\nTrain:", len(X_train))
print("Test :", len(X_test))


# ------------------------------------------------------------
# 11. StandardScaler
# Train으로만 fit
# ------------------------------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

X_all_scaled = scaler.transform(X)


# ------------------------------------------------------------
# 12. AutoEncoder
#
# 4 → 8 → 4 → 2 → 4 → 8 → 4
# ------------------------------------------------------------

input_layer = Input(shape=(4,))

x = Dense(
    8,
    activation="relu"
)(input_layer)

x = Dense(
    4,
    activation="relu"
)(x)

latent = Dense(
    2,
    activation="relu",
    name="Latent"
)(x)

x = Dense(
    4,
    activation="relu"
)(latent)

x = Dense(
    8,
    activation="relu"
)(x)

output_layer = Dense(
    4,
    activation="linear"
)(x)

autoencoder = Model(
    inputs=input_layer,
    outputs=output_layer
)

autoencoder.compile(
    optimizer="adam",
    loss="mse"
)

print("\n" + "=" * 80)
print("AutoEncoder")
print("=" * 80)

autoencoder.summary()


# ------------------------------------------------------------
# 13. 학습
# ------------------------------------------------------------

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

history = autoencoder.fit(
    X_train_scaled,
    X_train_scaled,

    validation_split=0.20,

    epochs=100,
    batch_size=32,

    shuffle=True,

    callbacks=[early_stop],

    verbose=1
)


# ------------------------------------------------------------
# 14. Loss 그래프
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Train Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("MSE Loss")

plt.title("AutoEncoder Training Loss")

plt.legend()
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 15. 전체 데이터 복원
# ------------------------------------------------------------

X_pred = autoencoder.predict(
    X_all_scaled,
    verbose=0
)


# ------------------------------------------------------------
# 16. 변수별 복원오차
# ------------------------------------------------------------

feature_errors = np.square(
    X_all_scaled - X_pred
)

data["Force_Error"] = feature_errors[:, 0]
data["Current_Error"] = feature_errors[:, 1]
data["Voltage_Error"] = feature_errors[:, 2]
data["Weld_Time_Error"] = feature_errors[:, 3]


# ------------------------------------------------------------
# 17. 행별 전체 복원오차
# ------------------------------------------------------------

reconstruction_error = np.mean(
    feature_errors,
    axis=1
)

data["Reconstruction_Error"] = reconstruction_error


print("\n" + "=" * 80)
print("전체 복원오차")
print("=" * 80)

print(
    data["Reconstruction_Error"].describe(
        percentiles=[
            0.50,
            0.90,
            0.95,
            0.99
        ]
    )
)


# ------------------------------------------------------------
# 18. 복원오차 분포
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.hist(
    reconstruction_error,
    bins=100
)

plt.xlabel("Reconstruction Error")
plt.ylabel("Count")

plt.title("Reconstruction Error Distribution")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 19. 복원오차를 3개로 KMeans
# ------------------------------------------------------------

error_X = reconstruction_error.reshape(-1, 1)

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=20
)

raw_cluster = kmeans.fit_predict(
    error_X
)


# ------------------------------------------------------------
# 20. 복원오차 순서대로 C0/C1/C2 재정렬
#
# C0 = 낮음
# C1 = 중간
# C2 = 높음
# ------------------------------------------------------------

centers = kmeans.cluster_centers_.flatten()

order = np.argsort(centers)

mapping = {
    order[0]: 0,
    order[1]: 1,
    order[2]: 2
}

data["AE_Cluster"] = np.array(
    [
        mapping[c]
        for c in raw_cluster
    ]
)


# ------------------------------------------------------------
# 21. Cluster 개수
# ------------------------------------------------------------

counts = (
    data["AE_Cluster"]
    .value_counts()
    .sort_index()
)

print("\n" + "=" * 80)
print("Cluster별 데이터 개수")
print("=" * 80)

for c in [0, 1, 2]:

    n = int(
        counts.get(c, 0)
    )

    ratio = (
        n / len(data) * 100
    )

    print(
        f"Cluster {c}: "
        f"{n}개 "
        f"({ratio:.2f}%)"
    )


# ------------------------------------------------------------
# 22. Cluster별 전체 복원오차
# ------------------------------------------------------------

cluster_summary = (
    data
    .groupby("AE_Cluster")
    ["Reconstruction_Error"]
    .agg(
        [
            "count",
            "min",
            "mean",
            "median",
            "max"
        ]
    )
)

print("\n" + "=" * 80)
print("Cluster별 전체 복원오차")
print("=" * 80)

print(
    cluster_summary.round(6)
)


# ------------------------------------------------------------
# 23. Silhouette
# ------------------------------------------------------------

sil = silhouette_score(
    error_X,
    data["AE_Cluster"]
)

print(
    "\nSilhouette Score:",
    round(sil, 4)
)


# ------------------------------------------------------------
# 24. 전체 행 그래프
# ------------------------------------------------------------

plt.figure(figsize=(16, 6))

for c in [0, 1, 2]:

    temp = data[
        data["AE_Cluster"] == c
    ]

    plt.scatter(
        temp["Original_Row"],
        temp["Reconstruction_Error"],
        s=15,
        label=f"Cluster {c}"
    )

plt.xlabel("Original Row Number")
plt.ylabel("Reconstruction Error")

plt.title(
    "AutoEncoder Reconstruction Error - 3 Clusters"
)

plt.legend()

plt.tight_layout()
plt.show()


# ============================================================
# 여기서부터 C1/C2 원인 분석
# Force 제외
# Current / Voltage / Weld Time만 분석
# ============================================================


# ------------------------------------------------------------
# 25. Cluster별 3개 변수 평균 복원오차
# ------------------------------------------------------------

analysis_errors = [
    "Current_Error",
    "Voltage_Error",
    "Weld_Time_Error"
]

mean_errors = (
    data
    .groupby("AE_Cluster")
    [analysis_errors]
    .mean()
)

print("\n" + "=" * 80)
print("Current / Voltage / Weld Time 평균 복원오차")
print("=" * 80)

print(
    mean_errors.round(6)
)


# ------------------------------------------------------------
# 26. 평균 복원오차 그래프
# ------------------------------------------------------------

mean_errors.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.xlabel("AE Cluster")
plt.ylabel("Mean Squared Reconstruction Error")

plt.title(
    "Feature Reconstruction Error by AE Cluster"
)

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 27. C1/C2만 추출
# ------------------------------------------------------------

candidate = data[
    data["AE_Cluster"].isin([1, 2])
].copy()


# ------------------------------------------------------------
# 28. C1/C2 각각의 행에서
# Current / Voltage / Weld Time 중
# 가장 복원오차가 큰 변수
# ------------------------------------------------------------

candidate["Largest_Error_Feature"] = (
    candidate[analysis_errors]
    .idxmax(axis=1)
    .replace(
        {
            "Current_Error": "Current",
            "Voltage_Error": "Voltage",
            "Weld_Time_Error": "Weld_Time"
        }
    )
)


# ------------------------------------------------------------
# 29. 가장 큰 오차 변수 개수
# ------------------------------------------------------------

dominant_count = pd.crosstab(
    candidate["AE_Cluster"],
    candidate["Largest_Error_Feature"]
)

print("\n" + "=" * 80)
print("C1/C2에서 가장 큰 복원오차 변수")
print("=" * 80)

print(dominant_count)


# ------------------------------------------------------------
# 30. 비율
# ------------------------------------------------------------

dominant_rate = (
    dominant_count
    .div(
        dominant_count.sum(axis=1),
        axis=0
    )
    * 100
)

print("\n" + "=" * 80)
print("C1/C2 가장 큰 복원오차 변수 비율 (%)")
print("=" * 80)

print(
    dominant_rate.round(2)
)


# ------------------------------------------------------------
# 31. 실제 공정값 비교
# ------------------------------------------------------------

actual_summary = (
    data
    .groupby("AE_Cluster")
    [
        [
            "Current",
            "Voltage",
            "Weld_Time"
        ]
    ]
    .agg(
        [
            "mean",
            "median",
            "std",
            "min",
            "max"
        ]
    )
)

print("\n" + "=" * 80)
print("Cluster별 실제 Current / Voltage / Weld Time")
print("=" * 80)

print(
    actual_summary.round(4)
)


# ------------------------------------------------------------
# 32. C1 결과
# ------------------------------------------------------------

c1 = candidate[
    candidate["AE_Cluster"] == 1
]

print("\n" + "=" * 80)
print("Cluster 1")
print("=" * 80)

print("행 개수:", len(c1))

print(
    "Current 평균 복원오차:",
    round(c1["Current_Error"].mean(), 6)
)

print(
    "Voltage 평균 복원오차:",
    round(c1["Voltage_Error"].mean(), 6)
)

print(
    "Weld Time 평균 복원오차:",
    round(c1["Weld_Time_Error"].mean(), 6)
)

print("\n가장 큰 오차 변수:")

print(
    c1["Largest_Error_Feature"]
    .value_counts()
)


# ------------------------------------------------------------
# 33. C2 결과
# ------------------------------------------------------------

c2 = candidate[
    candidate["AE_Cluster"] == 2
]

print("\n" + "=" * 80)
print("Cluster 2")
print("=" * 80)

print("행 개수:", len(c2))

print(
    "Current 평균 복원오차:",
    round(c2["Current_Error"].mean(), 6)
)

print(
    "Voltage 평균 복원오차:",
    round(c2["Voltage_Error"].mean(), 6)
)

print(
    "Weld Time 평균 복원오차:",
    round(c2["Weld_Time_Error"].mean(), 6)
)

print("\n가장 큰 오차 변수:")

print(
    c2["Largest_Error_Feature"]
    .value_counts()
)


# ------------------------------------------------------------
# 34. 결과 저장
# ------------------------------------------------------------

data.to_csv(
    "AE_weldtime_3cluster_result.csv",
    index=False,
    encoding="utf-8-sig"
)

candidate.to_csv(
    "AE_C1_C2_feature_analysis.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 80)
print("분석 완료")
print("=" * 80)

print("전체 데이터:", len(data))
print("C0:", (data["AE_Cluster"] == 0).sum())
print("C1:", (data["AE_Cluster"] == 1).sum())
print("C2:", (data["AE_Cluster"] == 2).sum())

print("\n실제 사용한 컬럼:")
print("Force     =", force_col)
print("Current   =", current_col)
print("Voltage   =", voltage_col)
print("Weld Time =", weld_time_col)

print("\n※ working time은 사용하지 않았습니다.")
```

```python
# ============================================================
# 3단계
# C1 / C2의 실제 값 + 날짜 + 원본 위치 분석
#
# 목적
# 1. C1/C2가 어느 날짜에 몰리는지 확인
# 2. C1을 Current / Voltage / Weld Time 유형으로 분리
# 3. C2가 어느 날짜에 몰리는지 확인
# 4. 실제 공정값과 원본 행 위치 확인
#
# 전제:
# 이전 셀의 AutoEncoder 분석이 실행되어
# data, df, candidate 변수가 존재해야 함
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# 1. 날짜 컬럼 찾기
#
# 기존 Raw data의 working time을
# AE 입력에는 사용하지 않았지만
# 여기서는 "날짜 확인용"으로만 사용
# ============================================================

time_candidates = []

for col in df.columns:

    name = (
        str(col)
        .lower()
        .replace("_", "")
        .replace(" ", "")
    )

    if "workingtime" in name:
        time_candidates.append(col)


if len(time_candidates) == 0:

    raise ValueError(
        "working time 컬럼을 찾지 못했습니다.\n"
        f"실제 컬럼: {df.columns.tolist()}"
    )


working_time_col = time_candidates[0]


print("=" * 80)
print("날짜 확인에 사용할 컬럼")
print("=" * 80)

print("Working Time =", working_time_col)


# ============================================================
# 2. 원본 working time을 data에 연결
#
# Original_Row를 이용하기 때문에
# AE 결과와 원본 데이터 위치를 맞춤
# ============================================================

data["Working_Time"] = (
    data["Original_Row"]
    .map(df[working_time_col])
)


# datetime 변환
data["Working_Time"] = pd.to_datetime(
    data["Working_Time"],
    errors="coerce"
)


# 날짜만 추출
data["Date"] = (
    data["Working_Time"]
    .dt.date
)


print("\n날짜 목록:")

print(
    sorted(
        data["Date"]
        .dropna()
        .unique()
    )
)


# ============================================================
# 3. C1/C2 다시 추출
# ============================================================

candidate = data[
    data["AE_Cluster"].isin([1, 2])
].copy()


# ============================================================
# 4. C1/C2 각각의 주요 복원오차 변수 결정
#
# Force는 여기서는 제외
#
# Current
# Voltage
# Weld Time
# ============================================================

error_cols = [
    "Current_Error",
    "Voltage_Error",
    "Weld_Time_Error"
]


candidate["Largest_Error_Feature"] = (
    candidate[error_cols]
    .idxmax(axis=1)
    .replace(
        {
            "Current_Error": "Current",
            "Voltage_Error": "Voltage",
            "Weld_Time_Error": "Weld_Time"
        }
    )
)


# ============================================================
# 5. 분석용 유형 이름 만들기
#
# 예:
# C1_Current
# C1_Voltage
# C1_Weld_Time
# C2_Voltage
# ============================================================

candidate["AE_Type"] = (
    "C"
    + candidate["AE_Cluster"].astype(str)
    + "_"
    + candidate["Largest_Error_Feature"]
)


print("\n" + "=" * 80)
print("AE 이상 후보 유형")
print("=" * 80)

print(
    candidate["AE_Type"]
    .value_counts()
)


# ============================================================
# 6. 날짜별 C1/C2 전체 개수
# ============================================================

daily_cluster = pd.crosstab(
    candidate["Date"],
    candidate["AE_Cluster"]
)


# 컬럼명 변경
daily_cluster = daily_cluster.rename(
    columns={
        1: "C1",
        2: "C2"
    }
)


# 없는 컬럼 방어
for col in ["C1", "C2"]:

    if col not in daily_cluster.columns:
        daily_cluster[col] = 0


daily_cluster["C1_C2_Total"] = (
    daily_cluster["C1"]
    + daily_cluster["C2"]
)


print("\n" + "=" * 80)
print("날짜별 C1 / C2 개수")
print("=" * 80)

print(daily_cluster)


# ============================================================
# 7. 날짜별 세부 유형
#
# C1_Current
# C1_Voltage
# C1_Weld_Time
# C2_Voltage 등
# ============================================================

daily_type = pd.crosstab(
    candidate["Date"],
    candidate["AE_Type"]
)


# 원하는 컬럼 순서
wanted_columns = [
    "C1_Current",
    "C1_Voltage",
    "C1_Weld_Time",
    "C2_Current",
    "C2_Voltage",
    "C2_Weld_Time"
]


for col in wanted_columns:

    if col not in daily_type.columns:
        daily_type[col] = 0


daily_type = daily_type[
    wanted_columns
]


daily_type["Total"] = (
    daily_type.sum(axis=1)
)


print("\n" + "=" * 80)
print("날짜별 AE 이상 후보 세부 유형")
print("=" * 80)

print(daily_type)


# ============================================================
# 8. 전체 날짜의 데이터 개수도 계산
#
# 단순 개수만 보면
# 날짜별 생산량 차이에 영향을 받을 수 있으므로
# 비율도 같이 계산
# ============================================================

daily_total = (
    data
    .groupby("Date")
    .size()
    .rename("Total_Rows")
)


daily_analysis = (
    daily_type
    .join(daily_total)
)


daily_analysis["Candidate_Rate_%"] = (
    daily_analysis["Total"]
    / daily_analysis["Total_Rows"]
    * 100
)


print("\n" + "=" * 80)
print("날짜별 전체 행 대비 C1/C2 비율")
print("=" * 80)

print(
    daily_analysis.round(2)
)


# ============================================================
# 9. C1만 날짜별 분석
# ============================================================

c1 = candidate[
    candidate["AE_Cluster"] == 1
].copy()


c1_daily = pd.crosstab(
    c1["Date"],
    c1["Largest_Error_Feature"]
)


for col in [
    "Current",
    "Voltage",
    "Weld_Time"
]:

    if col not in c1_daily.columns:
        c1_daily[col] = 0


c1_daily = c1_daily[
    [
        "Current",
        "Voltage",
        "Weld_Time"
    ]
]


c1_daily["Total"] = (
    c1_daily.sum(axis=1)
)


print("\n" + "=" * 80)
print("C1 날짜별 유형")
print("=" * 80)

print(c1_daily)


# ============================================================
# 10. C2만 날짜별 분석
# ============================================================

c2 = candidate[
    candidate["AE_Cluster"] == 2
].copy()


c2_daily = pd.crosstab(
    c2["Date"],
    c2["Largest_Error_Feature"]
)


print("\n" + "=" * 80)
print("C2 날짜별 유형")
print("=" * 80)

print(c2_daily)


# ============================================================
# 11. 날짜별 평균 실제 공정값
#
# C1/C2가 있는 날짜에서
# 실제 Current / Voltage / Weld Time 값 확인
# ============================================================

daily_actual = (
    candidate
    .groupby(
        [
            "Date",
            "AE_Cluster"
        ]
    )
    [
        [
            "Current",
            "Voltage",
            "Weld_Time"
        ]
    ]
    .agg(
        [
            "count",
            "mean",
            "median",
            "min",
            "max"
        ]
    )
)


print("\n" + "=" * 80)
print("날짜 × Cluster 실제 공정값")
print("=" * 80)

print(
    daily_actual.round(4)
)


# ============================================================
# 12. 원본 행 위치 확인
#
# 복원오차가 큰 순서대로
# ============================================================

detail_columns = [
    "Original_Row",
    "Date",
    "Working_Time",

    "AE_Cluster",
    "AE_Type",

    "Force",
    "Current",
    "Voltage",
    "Weld_Time",

    "Reconstruction_Error",

    "Current_Error",
    "Voltage_Error",
    "Weld_Time_Error"
]


candidate_detail = (
    candidate[
        detail_columns
    ]
    .sort_values(
        "Reconstruction_Error",
        ascending=False
    )
)


print("\n" + "=" * 80)
print("C1/C2 복원오차 상위 50개 원본 위치")
print("=" * 80)

print(
    candidate_detail
    .head(50)
    .to_string(index=False)
)


# ============================================================
# 13. 날짜별 그래프
# ============================================================

daily_type_plot = (
    daily_type[
        wanted_columns
    ]
)


daily_type_plot.plot(
    kind="bar",
    stacked=True,
    figsize=(14, 6)
)


plt.title(
    "Daily AE Candidate Types"
)

plt.xlabel(
    "Date"
)

plt.ylabel(
    "Number of Rows"
)

plt.xticks(
    rotation=45
)

plt.tight_layout()

plt.show()


# ============================================================
# 14. 날짜별 C1/C2 비율 그래프
# ============================================================

plt.figure(
    figsize=(14, 5)
)


plt.bar(
    daily_analysis.index.astype(str),
    daily_analysis["Candidate_Rate_%"]
)


plt.xlabel(
    "Date"
)

plt.ylabel(
    "C1 + C2 Rate (%)"
)

plt.title(
    "Daily AE Candidate Rate"
)

plt.xticks(
    rotation=45
)

plt.tight_layout()

plt.show()


# ============================================================
# 15. C2 위치만 따로 출력
#
# C2 = 현재 가장 높은 복원오차 군집
# ============================================================

c2_detail = (
    candidate_detail[
        candidate_detail["AE_Cluster"] == 2
    ]
    .sort_values(
        [
            "Date",
            "Original_Row"
        ]
    )
)


print("\n" + "=" * 80)
print("C2 전체 위치")
print("=" * 80)

print(
    c2_detail.to_string(index=False)
)


# ============================================================
# 16. 날짜별 C2 개수
# ============================================================

c2_date_count = (
    c2
    .groupby("Date")
    .size()
    .sort_values(
        ascending=False
    )
)


print("\n" + "=" * 80)
print("C2가 많이 발생한 날짜")
print("=" * 80)

print(c2_date_count)


# ============================================================
# 17. 저장
# ============================================================

daily_analysis.to_csv(
    "AE_daily_candidate_analysis.csv",
    encoding="utf-8-sig"
)


candidate_detail.to_csv(
    "AE_C1_C2_date_row_detail.csv",
    index=False,
    encoding="utf-8-sig"
)


c2_detail.to_csv(
    "AE_C2_detail.csv",
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 18. 최종 요약
# ============================================================

print("\n" + "=" * 80)
print("3단계 최종 요약")
print("=" * 80)


print(
    "\n전체 C1/C2:",
    len(candidate)
)


print(
    "\n유형별 개수:"
)

print(
    candidate[
        "AE_Type"
    ]
    .value_counts()
)


print(
    "\n날짜별 C1/C2:"
)

print(
    daily_analysis[
        [
            "Total",
            "Total_Rows",
            "Candidate_Rate_%"
        ]
    ]
    .round(2)
)


print(
    "\nC2 날짜별 개수:"
)

print(
    c2_date_count
)


print("\n저장 완료:")
print("AE_daily_candidate_analysis.csv")
print("AE_C1_C2_date_row_detail.csv")
print("AE_C2_detail.csv")
```

```python
# ============================================================
# 4단계
# AE C1/C2 ↔ 날짜별 반복 위치 ↔ Force 공정구조 비교
#
# 목적
# 1. 각 날짜 안에서 행 번호(Day_Row)를 새로 생성
# 2. C1/C2가 4개 날짜에서 같은 위치에 반복되는지 확인
# 3. Force가 낮은 구간 / 높은 구간에서 C1/C2가 어떻게 분포하는지 확인
# 4. 특히 C2 16개가 동일한 상대 위치/값으로 반복되는지 확인
#
# 주의:
# - C1/C2 = 실제 불량이라고 확정하지 않음
# - 현재는 "AE 이상 후보"로만 해석
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# 0. 이전 단계 결과 확인
# ============================================================

required_cols = [
    "Original_Row",
    "Date",
    "AE_Cluster",
    "Force",
    "Current",
    "Voltage",
    "Weld_Time",
    "Reconstruction_Error"
]

missing = [
    col for col in required_cols
    if col not in data.columns
]

if missing:
    raise ValueError(
        f"data에 필요한 컬럼이 없습니다: {missing}\n"
        "AutoEncoder + 3단계 코드를 먼저 실행하세요."
    )


# ============================================================
# 1. 원본 순서대로 정렬
# ============================================================

analysis = (
    data
    .copy()
    .sort_values("Original_Row")
    .reset_index(drop=True)
)


# ============================================================
# 2. 날짜 내부 상대 행 번호 생성
#
# 예:
# 3/25의 첫 행 = 1
# 3/25의 두 번째 행 = 2
# ...
#
# 날짜가 달라도 동일 위치 비교 가능
# ============================================================

analysis["Day_Row"] = (
    analysis
    .groupby("Date")
    .cumcount()
    + 1
)


# ============================================================
# 3. AE 후보 표시
# ============================================================

analysis["AE_Candidate"] = (
    analysis["AE_Cluster"]
    .isin([1, 2])
)


analysis["AE_Label"] = np.select(
    [
        analysis["AE_Cluster"] == 1,
        analysis["AE_Cluster"] == 2
    ],
    [
        "C1",
        "C2"
    ],
    default="C0"
)


# ============================================================
# 4. Force 기반 공정 상태 분류
#
# 기존 분석에서 관찰한 두 영역을 확인하기 위한 분류.
#
# 2.x대와 높은 Force 영역을 분리하되
# 애매한 값은 Other로 남김.
#
# 이것 자체가 실제 A/B 공정명이라고 확정하는 것은 아님.
# ============================================================

analysis["Force_Group"] = np.select(
    [
        (analysis["Force"] >= 2.0) &
        (analysis["Force"] < 3.0),

        analysis["Force"] >= 7.0
    ],
    [
        "Low_Force_2xx",
        "High_Force_7plus"
    ],
    default="Other"
)


# ============================================================
# 5. 날짜별 AE 후보 개수
# ============================================================

daily_ae = pd.crosstab(
    analysis["Date"],
    analysis["AE_Label"]
)


for col in ["C0", "C1", "C2"]:
    if col not in daily_ae.columns:
        daily_ae[col] = 0


daily_ae["C1+C2"] = (
    daily_ae["C1"]
    + daily_ae["C2"]
)


print("=" * 90)
print("날짜별 AE Cluster")
print("=" * 90)

print(daily_ae)


# ============================================================
# 6. C1/C2만 추출
# ============================================================

ae = analysis[
    analysis["AE_Candidate"]
].copy()


# ============================================================
# 7. 날짜 × Force Group × AE Cluster
#
# C1/C2가 Low Force / High Force 중
# 어디에 많이 존재하는지 확인
# ============================================================

force_ae = pd.crosstab(
    [
        ae["Date"],
        ae["Force_Group"]
    ],
    ae["AE_Label"]
)


for col in ["C1", "C2"]:
    if col not in force_ae.columns:
        force_ae[col] = 0


force_ae["Total"] = (
    force_ae["C1"]
    + force_ae["C2"]
)


print("\n" + "=" * 90)
print("날짜 × Force Group × AE 후보")
print("=" * 90)

print(force_ae)


# ============================================================
# 8. 전체 Force Group 대비 AE 후보 비율
#
# 단순 개수만 보면 해당 공정의 전체 행 수 차이 때문에
# 오해할 수 있으므로 비율도 계산
# ============================================================

force_total = (
    analysis
    .groupby(
        [
            "Date",
            "Force_Group"
        ]
    )
    .size()
    .rename("Total_Rows")
)


force_candidate = (
    ae
    .groupby(
        [
            "Date",
            "Force_Group"
        ]
    )
    .size()
    .rename("AE_Candidates")
)


force_rate = pd.concat(
    [
        force_total,
        force_candidate
    ],
    axis=1
).fillna(0)


force_rate["AE_Rate_%"] = (
    force_rate["AE_Candidates"]
    / force_rate["Total_Rows"]
    * 100
)


print("\n" + "=" * 90)
print("Force Group별 AE 후보 비율")
print("=" * 90)

print(force_rate.round(2))


# ============================================================
# 9. 핵심:
# 날짜 내부의 동일 위치에서 C1/C2가 반복되는지 확인
#
# Day_Row 기준으로
# 몇 개 날짜에서 C1/C2가 등장했는지 계산
# ============================================================

position_repeat = (
    ae
    .groupby(
        [
            "Day_Row",
            "AE_Label"
        ]
    )
    .agg(
        Repeat_Dates=("Date", "nunique"),
        Total_Count=("Date", "size"),
        Mean_Force=("Force", "mean"),
        Mean_Current=("Current", "mean"),
        Mean_Voltage=("Voltage", "mean"),
        Mean_Weld_Time=("Weld_Time", "mean"),
        Mean_AE_Error=("Reconstruction_Error", "mean")
    )
    .reset_index()
)


position_repeat = (
    position_repeat
    .sort_values(
        [
            "Repeat_Dates",
            "AE_Label",
            "Day_Row"
        ],
        ascending=[
            False,
            True,
            True
        ]
    )
)


print("\n" + "=" * 90)
print("동일 날짜 내부 위치에서 반복되는 AE 후보")
print("=" * 90)

print(
    position_repeat
    .head(100)
    .round(4)
    .to_string(index=False)
)


# ============================================================
# 10. 4개 날짜 모두에서 반복된 위치
#
# 현재 결과에서 주요 반복 날짜가
# 3/25, 3/27, 3/31, 4/3이므로
# 4회 반복 위치를 확인
# ============================================================

repeat_4 = position_repeat[
    position_repeat["Repeat_Dates"] >= 4
].copy()


print("\n" + "=" * 90)
print("4개 이상 날짜에서 동일하게 반복된 AE 위치")
print("=" * 90)

print(
    repeat_4
    .round(4)
    .to_string(index=False)
)


print(
    "\n반복 위치 개수:",
    len(repeat_4)
)


# ============================================================
# 11. C1 / C2 각각 반복 위치 개수
# ============================================================

repeat_cluster_count = (
    repeat_4["AE_Label"]
    .value_counts()
)


print("\n" + "=" * 90)
print("반복 위치의 Cluster 구성")
print("=" * 90)

print(repeat_cluster_count)


# ============================================================
# 12. C2만 집중 분석
# ============================================================

c2 = ae[
    ae["AE_Cluster"] == 2
].copy()


print("\n" + "=" * 90)
print("C2 날짜별 위치")
print("=" * 90)

for date, group in c2.groupby("Date"):

    print(f"\n[{date}]")

    print(
        group[
            [
                "Day_Row",
                "Original_Row",
                "Force",
                "Force_Group",
                "Current",
                "Voltage",
                "Weld_Time",
                "Reconstruction_Error"
            ]
        ]
        .sort_values("Day_Row")
        .round(4)
        .to_string(index=False)
    )


# ============================================================
# 13. 날짜별 C2 Day_Row 비교표
#
# 같은 상대 위치에서 발생하는지 직접 확인
# ============================================================

c2_position_table = (
    c2
    .groupby("Date")["Day_Row"]
    .apply(list)
)


print("\n" + "=" * 90)
print("날짜별 C2 상대 위치")
print("=" * 90)

for date, positions in c2_position_table.items():

    print(
        date,
        "→",
        positions
    )


# ============================================================
# 14. C2의 상대 위치가 날짜마다 완전히 같은지 검사
# ============================================================

position_lists = [
    tuple(x)
    for x in c2_position_table
]


if len(position_lists) > 0:

    same_c2_positions = (
        len(set(position_lists)) == 1
    )

else:

    same_c2_positions = False


print("\n" + "=" * 90)
print("C2 위치 반복성 검사")
print("=" * 90)

print(
    "각 날짜 C2 위치가 완전히 동일한가? →",
    same_c2_positions
)


# ============================================================
# 15. C2 실제 공정값 패턴 비교
#
# Force / Current / Voltage / Weld Time의
# 순서까지 동일한지 확인
# ============================================================

pattern_cols = [
    "Force",
    "Current",
    "Voltage",
    "Weld_Time"
]


c2_patterns = {}


for date, group in c2.groupby("Date"):

    temp = (
        group
        .sort_values("Day_Row")
        [pattern_cols]
        .round(6)
    )

    c2_patterns[date] = tuple(
        map(
            tuple,
            temp.to_numpy()
        )
    )


if len(c2_patterns) > 0:

    same_c2_values = (
        len(
            set(
                c2_patterns.values()
            )
        )
        == 1
    )

else:

    same_c2_values = False


print("\nC2의 Force/Current/Voltage/Weld Time 순서가")
print(
    "각 날짜에서 완전히 동일한가? →",
    same_c2_values
)


# ============================================================
# 16. 연속된 AE 후보 블록 찾기
#
# C1/C2가 낱개로 발생하는지
# 아니면 연속적인 공정 블록으로 발생하는지 확인
# ============================================================

blocks = []


for date, group in ae.groupby("Date"):

    group = (
        group
        .sort_values("Day_Row")
        .copy()
    )

    positions = (
        group["Day_Row"]
        .to_numpy()
    )

    if len(positions) == 0:
        continue


    start = positions[0]
    previous = positions[0]


    for pos in positions[1:]:

        if pos == previous + 1:

            previous = pos

        else:

            blocks.append(
                [
                    date,
                    start,
                    previous,
                    previous - start + 1
                ]
            )

            start = pos
            previous = pos


    blocks.append(
        [
            date,
            start,
            previous,
            previous - start + 1
        ]
    )


block_df = pd.DataFrame(
    blocks,
    columns=[
        "Date",
        "Start_Day_Row",
        "End_Day_Row",
        "Block_Length"
    ]
)


block_df = (
    block_df
    .sort_values(
        [
            "Block_Length",
            "Date"
        ],
        ascending=[
            False,
            True
        ]
    )
)


print("\n" + "=" * 90)
print("C1/C2 연속 발생 블록")
print("=" * 90)

print(
    block_df
    .head(100)
    .to_string(index=False)
)


# ============================================================
# 17. 반복되는 블록 자체 확인
#
# Start / End / Length가 여러 날짜에서
# 동일하게 나타나는지 확인
# ============================================================

repeated_blocks = (
    block_df
    .groupby(
        [
            "Start_Day_Row",
            "End_Day_Row",
            "Block_Length"
        ]
    )
    .agg(
        Repeat_Dates=("Date", "nunique"),
        Dates=(
            "Date",
            lambda x: ", ".join(
                map(
                    str,
                    sorted(set(x))
                )
            )
        )
    )
    .reset_index()
    .sort_values(
        [
            "Repeat_Dates",
            "Block_Length"
        ],
        ascending=[
            False,
            False
        ]
    )
)


print("\n" + "=" * 90)
print("여러 날짜에서 반복된 AE 연속 블록")
print("=" * 90)

print(
    repeated_blocks
    .head(100)
    .to_string(index=False)
)


# ============================================================
# 18. C1/C2 위치 그래프
#
# X = 날짜 내부 상대 행
# Y = 복원오차
# ============================================================

plt.figure(
    figsize=(16, 6)
)


for cluster in [1, 2]:

    temp = ae[
        ae["AE_Cluster"] == cluster
    ]

    plt.scatter(
        temp["Day_Row"],
        temp["Reconstruction_Error"],
        s=18,
        alpha=0.65,
        label=f"C{cluster}"
    )


plt.xlabel(
    "Row Position Within Date"
)

plt.ylabel(
    "Reconstruction Error"
)

plt.title(
    "Repeated AE Candidate Positions Within Each Date"
)

plt.legend()

plt.grid(
    alpha=0.2
)

plt.tight_layout()

plt.show()


# ============================================================
# 19. 날짜별 그래프
#
# 날짜마다 따로 그려서
# 동일 패턴인지 눈으로 확인
# ============================================================

important_dates = sorted(
    ae["Date"]
    .dropna()
    .unique()
)


for date in important_dates:

    temp_all = analysis[
        analysis["Date"] == date
    ]

    temp_ae = ae[
        ae["Date"] == date
    ]


    plt.figure(
        figsize=(15, 4)
    )


    # Force 전체 구조
    plt.plot(
        temp_all["Day_Row"],
        temp_all["Force"],
        alpha=0.4,
        label="Force"
    )


    # C1
    c1_temp = temp_ae[
        temp_ae["AE_Cluster"] == 1
    ]

    plt.scatter(
        c1_temp["Day_Row"],
        c1_temp["Force"],
        s=20,
        label="C1"
    )


    # C2
    c2_temp = temp_ae[
        temp_ae["AE_Cluster"] == 2
    ]

    plt.scatter(
        c2_temp["Day_Row"],
        c2_temp["Force"],
        s=35,
        label="C2"
    )


    plt.title(
        f"{date} - Force Structure and AE Candidates"
    )

    plt.xlabel(
        "Row Position Within Date"
    )

    plt.ylabel(
        "Force"
    )

    plt.legend()

    plt.grid(
        alpha=0.2
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 20. 저장
# ============================================================

force_rate.to_csv(
    "AE_force_group_comparison.csv",
    encoding="utf-8-sig"
)


position_repeat.to_csv(
    "AE_repeated_positions.csv",
    index=False,
    encoding="utf-8-sig"
)


repeat_4.to_csv(
    "AE_positions_repeated_4dates.csv",
    index=False,
    encoding="utf-8-sig"
)


block_df.to_csv(
    "AE_continuous_blocks.csv",
    index=False,
    encoding="utf-8-sig"
)


repeated_blocks.to_csv(
    "AE_repeated_blocks.csv",
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 21. 최종 요약
# ============================================================

print("\n" + "=" * 90)
print("4단계 최종 요약")
print("=" * 90)


print(
    "전체 AE 후보:",
    len(ae)
)


print(
    "\nC1:",
    (ae["AE_Cluster"] == 1).sum()
)


print(
    "C2:",
    (ae["AE_Cluster"] == 2).sum()
)


print(
    "\n4개 이상 날짜에서 반복된 AE 상대 위치:",
    len(repeat_4)
)


print(
    "\nC2 위치가 날짜마다 완전히 동일:",
    same_c2_positions
)


print(
    "C2 공정값 패턴이 날짜마다 완전히 동일:",
    same_c2_values
)


print("\nForce Group별 AE 후보:")

print(
    ae["Force_Group"]
    .value_counts()
)


print("\n저장 파일:")
print("1. AE_force_group_comparison.csv")
print("2. AE_repeated_positions.csv")
print("3. AE_positions_repeated_4dates.csv")
print("4. AE_continuous_blocks.csv")
print("5. AE_repeated_blocks.csv")
```

```python
# ============================================================
# 5단계
# AE C1/C2 오류 후보를 "변수 조합"에 따라 세분화
#
# 대상:
#   C1 = 997개
#   C2 = 64개
#
# 사용:
#   Current_Error
#   Voltage_Error
#   Weld_Time_Error
#
# 제외:
#   Force
#
# 목적:
#   Current 단독
#   Voltage 단독
#   Weld Time 단독
#   Current + Voltage
#   Current + Weld Time
#   Voltage + Weld Time
#   Current + Voltage + Weld Time
#
# 주의:
#   여기서 "오류 유형"은 실제 불량 종류가 아니라
#   AE 복원오차 패턴에 따른 후보 유형임
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# 1. 필요한 데이터 확인
# ============================================================

required_cols = [
    "AE_Cluster",
    "Current_Error",
    "Voltage_Error",
    "Weld_Time_Error",
    "Current",
    "Voltage",
    "Weld_Time",
    "Reconstruction_Error"
]

missing = [
    col for col in required_cols
    if col not in data.columns
]

if missing:
    raise ValueError(
        f"다음 컬럼이 없습니다: {missing}\n"
        "앞의 AutoEncoder 코드를 먼저 실행하세요."
    )


# ============================================================
# 2. C1/C2만 추출
# ============================================================

candidate = data[
    data["AE_Cluster"].isin([1, 2])
].copy()


print("=" * 90)
print("분석 대상")
print("=" * 90)

print("전체 후보:", len(candidate))
print("C1:", (candidate["AE_Cluster"] == 1).sum())
print("C2:", (candidate["AE_Cluster"] == 2).sum())


# ============================================================
# 3. 변수별 복원오차 컬럼
# ============================================================

error_cols = [
    "Current_Error",
    "Voltage_Error",
    "Weld_Time_Error"
]


# ============================================================
# 4. 각 변수의 복원오차를 행 내부 비율로 변환
#
# 예:
#
# Current Error = 5
# Voltage Error = 8
# Time Error    = 1
#
# 합계 = 14
#
# Current 비중 = 35.7%
# Voltage 비중 = 57.1%
# Time 비중    = 7.1%
#
# 이렇게 하면 단순히 "가장 큰 변수 하나"가 아니라
# 동시에 크게 작용한 변수도 찾을 수 있음
# ============================================================

candidate["Selected_Error_Sum"] = (
    candidate["Current_Error"]
    + candidate["Voltage_Error"]
    + candidate["Weld_Time_Error"]
)


# 0으로 나누는 문제 방지
denominator = (
    candidate["Selected_Error_Sum"]
    .replace(0, np.nan)
)


candidate["Current_Error_Ratio"] = (
    candidate["Current_Error"]
    / denominator
)


candidate["Voltage_Error_Ratio"] = (
    candidate["Voltage_Error"]
    / denominator
)


candidate["Weld_Time_Error_Ratio"] = (
    candidate["Weld_Time_Error"]
    / denominator
)


# ============================================================
# 5. "주요 오류 변수" 기준
#
# 한 변수의 복원오차가
# 세 변수 복원오차 합의 30% 이상이면
# 해당 행에서 주요 변수로 포함
#
# 30%는 분석용 기준이지
# 실제 불량을 판정하는 절대 기준이 아님.
# ============================================================

RATIO_THRESHOLD = 0.30


candidate["Current_High"] = (
    candidate["Current_Error_Ratio"]
    >= RATIO_THRESHOLD
)

candidate["Voltage_High"] = (
    candidate["Voltage_Error_Ratio"]
    >= RATIO_THRESHOLD
)

candidate["Weld_Time_High"] = (
    candidate["Weld_Time_Error_Ratio"]
    >= RATIO_THRESHOLD
)


# ============================================================
# 6. 오류 조합 분류 함수
# ============================================================

def classify_error_type(row):

    features = []

    if row["Current_High"]:
        features.append("Current")

    if row["Voltage_High"]:
        features.append("Voltage")

    if row["Weld_Time_High"]:
        features.append("Weld_Time")


    # 30% 이상인 변수가 하나도 없을 가능성에 대비
    if len(features) == 0:

        values = {
            "Current": row["Current_Error"],
            "Voltage": row["Voltage_Error"],
            "Weld_Time": row["Weld_Time_Error"]
        }

        return max(
            values,
            key=values.get
        )


    return " + ".join(features)


candidate["AE_Error_Type"] = (
    candidate.apply(
        classify_error_type,
        axis=1
    )
)


# ============================================================
# 7. 전체 오류 유형 개수
# ============================================================

type_count = (
    candidate["AE_Error_Type"]
    .value_counts()
)


type_rate = (
    candidate["AE_Error_Type"]
    .value_counts(normalize=True)
    * 100
)


type_summary = pd.DataFrame({
    "Count": type_count,
    "Rate_%": type_rate
})


print("\n" + "=" * 90)
print("C1/C2 오류 후보 유형")
print("=" * 90)

print(
    type_summary
    .round(2)
)


# ============================================================
# 8. C1과 C2를 따로 분석
# ============================================================

cluster_type = pd.crosstab(
    candidate["AE_Cluster"],
    candidate["AE_Error_Type"]
)


print("\n" + "=" * 90)
print("C1 / C2별 오류 유형 개수")
print("=" * 90)

print(cluster_type)


# ============================================================
# 9. C1/C2 내부 비율
# ============================================================

cluster_type_rate = pd.crosstab(
    candidate["AE_Cluster"],
    candidate["AE_Error_Type"],
    normalize="index"
) * 100


print("\n" + "=" * 90)
print("C1 / C2별 오류 유형 비율 (%)")
print("=" * 90)

print(
    cluster_type_rate
    .round(2)
)


# ============================================================
# 10. 변수별 평균 복원오차 비중
# ============================================================

ratio_summary = (
    candidate
    .groupby("AE_Cluster")
    [
        [
            "Current_Error_Ratio",
            "Voltage_Error_Ratio",
            "Weld_Time_Error_Ratio"
        ]
    ]
    .mean()
    * 100
)


print("\n" + "=" * 90)
print("Cluster별 평균 복원오차 기여 비율 (%)")
print("=" * 90)

print(
    ratio_summary
    .round(2)
)


# ============================================================
# 11. 오류 유형별 실제 공정값
#
# 여기서 실제 Current / Voltage / Weld Time이
# 어떤 값을 가지는지 확인
# ============================================================

actual_value_summary = (
    candidate
    .groupby("AE_Error_Type")
    [
        [
            "Current",
            "Voltage",
            "Weld_Time"
        ]
    ]
    .agg(
        [
            "count",
            "mean",
            "median",
            "std",
            "min",
            "max"
        ]
    )
)


print("\n" + "=" * 90)
print("오류 유형별 실제 공정값")
print("=" * 90)

print(
    actual_value_summary
    .round(4)
)


# ============================================================
# 12. 오류 유형별 평균 복원오차
# ============================================================

error_value_summary = (
    candidate
    .groupby("AE_Error_Type")
    [
        [
            "Current_Error",
            "Voltage_Error",
            "Weld_Time_Error",
            "Reconstruction_Error"
        ]
    ]
    .mean()
)


print("\n" + "=" * 90)
print("오류 유형별 평균 복원오차")
print("=" * 90)

print(
    error_value_summary
    .round(4)
)


# ============================================================
# 13. C2만 따로 확인
# ============================================================

c2 = candidate[
    candidate["AE_Cluster"] == 2
].copy()


print("\n" + "=" * 90)
print("C2 오류 유형")
print("=" * 90)

print(
    c2["AE_Error_Type"]
    .value_counts()
)


print("\nC2 평균 복원오차 비중 (%)")

if len(c2) > 0:

    print(
        (
            c2[
                [
                    "Current_Error_Ratio",
                    "Voltage_Error_Ratio",
                    "Weld_Time_Error_Ratio"
                ]
            ]
            .mean()
            * 100
        )
        .round(2)
    )


# ============================================================
# 14. C1만 따로 확인
# ============================================================

c1 = candidate[
    candidate["AE_Cluster"] == 1
].copy()


print("\n" + "=" * 90)
print("C1 오류 유형")
print("=" * 90)

print(
    c1["AE_Error_Type"]
    .value_counts()
)


# ============================================================
# 15. 각 행의 상세 결과
# ============================================================

detail_cols = []


# 이전 단계에서 존재할 수도 있는 컬럼
for col in [
    "Original_Row",
    "Date",
    "Working_Time"
]:

    if col in candidate.columns:
        detail_cols.append(col)


detail_cols += [

    "AE_Cluster",
    "AE_Error_Type",

    "Current",
    "Voltage",
    "Weld_Time",

    "Reconstruction_Error",

    "Current_Error",
    "Voltage_Error",
    "Weld_Time_Error",

    "Current_Error_Ratio",
    "Voltage_Error_Ratio",
    "Weld_Time_Error_Ratio"
]


candidate_detail = (
    candidate[
        detail_cols
    ]
    .sort_values(
        "Reconstruction_Error",
        ascending=False
    )
)


print("\n" + "=" * 90)
print("복원오차가 큰 상위 50행")
print("=" * 90)

print(
    candidate_detail
    .head(50)
    .round(4)
    .to_string(index=False)
)


# ============================================================
# 16. 오류 유형 그래프
# ============================================================

plot_count = (
    candidate["AE_Error_Type"]
    .value_counts()
)


plt.figure(
    figsize=(12, 6)
)


plt.bar(
    plot_count.index,
    plot_count.values
)


plt.xlabel(
    "AE Error Type"
)

plt.ylabel(
    "Number of Rows"
)

plt.title(
    "AE C1/C2 Error Pattern Classification"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 17. C1 / C2 유형 비교 그래프
# ============================================================

cluster_type.T.plot(
    kind="bar",
    figsize=(12, 6)
)


plt.xlabel(
    "AE Error Type"
)

plt.ylabel(
    "Number of Rows"
)

plt.title(
    "Error Pattern by AE Cluster"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 18. 저장
# ============================================================

type_summary.to_csv(
    "AE_error_type_summary.csv",
    encoding="utf-8-sig"
)


cluster_type.to_csv(
    "AE_cluster_error_type_count.csv",
    encoding="utf-8-sig"
)


cluster_type_rate.to_csv(
    "AE_cluster_error_type_rate.csv",
    encoding="utf-8-sig"
)


candidate_detail.to_csv(
    "AE_C1_C2_error_type_detail.csv",
    index=False,
    encoding="utf-8-sig"
)


actual_value_summary.to_csv(
    "AE_error_type_actual_values.csv",
    encoding="utf-8-sig"
)


# ============================================================
# 19. 최종 결과
# ============================================================

print("\n" + "=" * 90)
print("5단계 최종 결과")
print("=" * 90)

print(
    f"분류 기준: 각 행에서 복원오차 기여율 "
    f"{RATIO_THRESHOLD*100:.0f}% 이상"
)

print(
    "\n전체 C1/C2 후보:",
    len(candidate)
)

print(
    "\n오류 패턴별 개수:"
)

print(type_count)


print(
    "\n오류 패턴별 비율(%):"
)

print(
    type_rate.round(2)
)


print(
    "\n※ 이 분류는 실제 불량 유형이 아니라 "
    "AE 복원오차의 변수 조합에 따른 후보 유형입니다."
)
```

```python
# ============================================================
# 6단계
# AE 오류 후보의 실제 공정값 "방향" 분석
#
# 목적
# ------------------------------------------------------------
# C1/C2의 복원오차가 큰 이유를
# 실제 공정값의 방향과 연결
#
# 예:
# Voltage 오류 후보
#     ↓
# 실제 Voltage가 C0보다 낮은가?
# 실제 Voltage가 C0보다 높은가?
#
# Current / Weld Time도 동일
#
# 기준:
# C0의 각 변수 분포
#
# 주의:
# C0 = 실제 정상이라고 확정하는 것이 아니라
# AE상 복원오차가 낮은 기준군으로 사용
# ============================================================


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# 1. 필요한 컬럼 확인
# ============================================================

required_cols = [
    "AE_Cluster",

    "Current",
    "Voltage",
    "Weld_Time",

    "Current_Error",
    "Voltage_Error",
    "Weld_Time_Error",

    "Reconstruction_Error"
]


missing = [
    col for col in required_cols
    if col not in data.columns
]


if missing:

    raise ValueError(
        f"필요한 컬럼이 없습니다: {missing}\n"
        "앞의 AutoEncoder 분석을 먼저 실행하세요."
    )


# ============================================================
# 2. C0 / C1 / C2 분리
# ============================================================

c0 = data[
    data["AE_Cluster"] == 0
].copy()


candidate = data[
    data["AE_Cluster"].isin([1, 2])
].copy()


print("=" * 90)
print("분석 데이터")
print("=" * 90)

print("C0:", len(c0))
print("C1:", (candidate["AE_Cluster"] == 1).sum())
print("C2:", (candidate["AE_Cluster"] == 2).sum())
print("C1+C2:", len(candidate))


# ============================================================
# 3. C0 기준값 계산
#
# 평균 / 중앙값 / 표준편차 / 분위수
# ============================================================

features = [
    "Current",
    "Voltage",
    "Weld_Time"
]


baseline = pd.DataFrame(
    index=features
)


baseline["mean"] = [
    c0[x].mean()
    for x in features
]


baseline["median"] = [
    c0[x].median()
    for x in features
]


baseline["std"] = [
    c0[x].std()
    for x in features
]


baseline["Q05"] = [
    c0[x].quantile(0.05)
    for x in features
]


baseline["Q25"] = [
    c0[x].quantile(0.25)
    for x in features
]


baseline["Q75"] = [
    c0[x].quantile(0.75)
    for x in features
]


baseline["Q95"] = [
    c0[x].quantile(0.95)
    for x in features
]


print("\n" + "=" * 90)
print("C0 기준 공정값")
print("=" * 90)

print(
    baseline.round(4)
)


# ============================================================
# 4. 실제 값의 방향 계산
#
# 우선 C0 중앙값을 중심으로
#
# < 중앙값 → Low
# > 중앙값 → High
#
# 로 구분
# ============================================================

for feature in features:

    median_value = baseline.loc[
        feature,
        "median"
    ]

    candidate[
        f"{feature}_Direction"
    ] = np.where(
        candidate[feature] < median_value,
        "Low",
        np.where(
            candidate[feature] > median_value,
            "High",
            "Same"
        )
    )


# ============================================================
# 5. C0의 5~95% 범위를 이용한
# 보다 강한 방향 분류
#
# Low_Extreme  = C0 하위 5%보다 낮음
# Normal_Range = C0 5~95% 범위
# High_Extreme = C0 상위 5%보다 높음
#
# 여기서 Extreme도 "실제 오류"라는 뜻은 아님.
# ============================================================

for feature in features:

    low = baseline.loc[
        feature,
        "Q05"
    ]

    high = baseline.loc[
        feature,
        "Q95"
    ]


    candidate[
        f"{feature}_Range"
    ] = np.select(

        [
            candidate[feature] < low,
            candidate[feature] > high
        ],

        [
            "Low_Extreme",
            "High_Extreme"
        ],

        default="C0_5_95_Range"
    )


# ============================================================
# 6. 이전 5단계의 AE_Error_Type이 없으면 다시 생성
# ============================================================

if "AE_Error_Type" not in candidate.columns:

    candidate["Selected_Error_Sum"] = (
        candidate["Current_Error"]
        + candidate["Voltage_Error"]
        + candidate["Weld_Time_Error"]
    )


    denominator = (
        candidate["Selected_Error_Sum"]
        .replace(0, np.nan)
    )


    candidate["Current_Error_Ratio"] = (
        candidate["Current_Error"]
        / denominator
    )


    candidate["Voltage_Error_Ratio"] = (
        candidate["Voltage_Error"]
        / denominator
    )


    candidate["Weld_Time_Error_Ratio"] = (
        candidate["Weld_Time_Error"]
        / denominator
    )


    RATIO_THRESHOLD = 0.30


    def classify_error_type(row):

        selected = []


        if (
            row["Current_Error_Ratio"]
            >= RATIO_THRESHOLD
        ):

            selected.append(
                "Current"
            )


        if (
            row["Voltage_Error_Ratio"]
            >= RATIO_THRESHOLD
        ):

            selected.append(
                "Voltage"
            )


        if (
            row["Weld_Time_Error_Ratio"]
            >= RATIO_THRESHOLD
        ):

            selected.append(
                "Weld_Time"
            )


        if len(selected) == 0:

            values = {
                "Current":
                    row["Current_Error"],

                "Voltage":
                    row["Voltage_Error"],

                "Weld_Time":
                    row["Weld_Time_Error"]
            }


            return max(
                values,
                key=values.get
            )


        return " + ".join(
            selected
        )


    candidate["AE_Error_Type"] = (
        candidate.apply(
            classify_error_type,
            axis=1
        )
    )


# ============================================================
# 7. Voltage 관련 후보
# ============================================================

voltage_candidate = candidate[
    candidate["AE_Error_Type"]
    .str.contains(
        "Voltage",
        na=False
    )
].copy()


print("\n" + "=" * 90)
print("Voltage 관련 AE 후보")
print("=" * 90)

print(
    "개수:",
    len(voltage_candidate)
)


print(
    "\nC0 중앙값 기준 방향:"
)

print(
    voltage_candidate[
        "Voltage_Direction"
    ]
    .value_counts()
)


print(
    "\nC0 5~95% 범위 기준:"
)

print(
    voltage_candidate[
        "Voltage_Range"
    ]
    .value_counts()
)


# ============================================================
# 8. Current 관련 후보
# ============================================================

current_candidate = candidate[
    candidate["AE_Error_Type"]
    .str.contains(
        "Current",
        na=False
    )
].copy()


print("\n" + "=" * 90)
print("Current 관련 AE 후보")
print("=" * 90)

print(
    "개수:",
    len(current_candidate)
)


print(
    "\nC0 중앙값 기준 방향:"
)

print(
    current_candidate[
        "Current_Direction"
    ]
    .value_counts()
)


print(
    "\nC0 5~95% 범위 기준:"
)

print(
    current_candidate[
        "Current_Range"
    ]
    .value_counts()
)


# ============================================================
# 9. Weld Time 관련 후보
# ============================================================

time_candidate = candidate[
    candidate["AE_Error_Type"]
    .str.contains(
        "Weld_Time",
        na=False
    )
].copy()


print("\n" + "=" * 90)
print("Weld Time 관련 AE 후보")
print("=" * 90)

print(
    "개수:",
    len(time_candidate)
)


print(
    "\nC0 중앙값 기준 방향:"
)

print(
    time_candidate[
        "Weld_Time_Direction"
    ]
    .value_counts()
)


print(
    "\nC0 5~95% 범위 기준:"
)

print(
    time_candidate[
        "Weld_Time_Range"
    ]
    .value_counts()
)


# ============================================================
# 10. AE 유형 + 실제 방향 결합
#
# 예:
#
# Voltage_Low
# Voltage_High
#
# Current + Voltage에서
# Current_High + Voltage_Low
#
# 형태로 만듦
# ============================================================

def make_direction_pattern(row):

    result = []


    if "Current" in row["AE_Error_Type"]:

        result.append(
            "Current_"
            + row["Current_Direction"]
        )


    if "Voltage" in row["AE_Error_Type"]:

        result.append(
            "Voltage_"
            + row["Voltage_Direction"]
        )


    if "Weld_Time" in row["AE_Error_Type"]:

        result.append(
            "WeldTime_"
            + row["Weld_Time_Direction"]
        )


    return " + ".join(
        result
    )


candidate["Direction_Pattern"] = (
    candidate.apply(
        make_direction_pattern,
        axis=1
    )
)


# ============================================================
# 11. 실제 방향 패턴 개수
# ============================================================

direction_count = (
    candidate[
        "Direction_Pattern"
    ]
    .value_counts()
)


direction_rate = (
    candidate[
        "Direction_Pattern"
    ]
    .value_counts(
        normalize=True
    )
    * 100
)


direction_summary = pd.DataFrame({

    "Count":
        direction_count,

    "Rate_%":
        direction_rate
})


print("\n" + "=" * 90)
print("실제 공정값 방향까지 결합한 AE 유형")
print("=" * 90)

print(
    direction_summary
    .round(2)
)


# ============================================================
# 12. C1 / C2별 방향 패턴
# ============================================================

cluster_direction = pd.crosstab(
    candidate["AE_Cluster"],
    candidate["Direction_Pattern"]
)


print("\n" + "=" * 90)
print("C1 / C2별 실제 방향 패턴")
print("=" * 90)

print(
    cluster_direction
)


# ============================================================
# 13. C1 / C2 실제 값 비교
# ============================================================

cluster_actual = (
    data
    .groupby(
        "AE_Cluster"
    )
    [
        [
            "Current",
            "Voltage",
            "Weld_Time"
        ]
    ]
    .agg(
        [
            "count",
            "mean",
            "median",
            "std",
            "min",
            "max"
        ]
    )
)


print("\n" + "=" * 90)
print("C0 / C1 / C2 실제 공정값 비교")
print("=" * 90)

print(
    cluster_actual
    .round(4)
)


# ============================================================
# 14. 각 AE 오류 유형별 실제 평균값
# ============================================================

type_actual = (
    candidate
    .groupby(
        "AE_Error_Type"
    )
    [
        [
            "Current",
            "Voltage",
            "Weld_Time"
        ]
    ]
    .agg(
        [
            "count",
            "mean",
            "median",
            "std",
            "min",
            "max"
        ]
    )
)


print("\n" + "=" * 90)
print("AE 오류 유형별 실제 공정값")
print("=" * 90)

print(
    type_actual
    .round(4)
)


# ============================================================
# 15. Extreme 조합 생성
#
# C0 5~95% 범위를 실제로 벗어난 변수만 표시
#
# 예:
# Voltage_Low_Extreme
# Current_High_Extreme
# Current_High_Extreme + Voltage_Low_Extreme
# ============================================================

def make_extreme_pattern(row):

    result = []


    for feature in features:

        value = row[
            f"{feature}_Range"
        ]


        if value == "Low_Extreme":

            result.append(
                feature
                + "_Low_Extreme"
            )


        elif value == "High_Extreme":

            result.append(
                feature
                + "_High_Extreme"
            )


    if len(result) == 0:

        return "Within_C0_5_95"


    return " + ".join(
        result
    )


candidate["Extreme_Pattern"] = (
    candidate.apply(
        make_extreme_pattern,
        axis=1
    )
)


extreme_count = (
    candidate[
        "Extreme_Pattern"
    ]
    .value_counts()
)


print("\n" + "=" * 90)
print("C0 5~95% 범위를 벗어난 실제 공정값 조합")
print("=" * 90)

print(
    extreme_count
)


# ============================================================
# 16. 복원오차가 큰 순서로 상세 확인
# ============================================================

detail_columns = []


for col in [
    "Original_Row",
    "Date",
    "Working_Time"
]:

    if col in candidate.columns:

        detail_columns.append(
            col
        )


detail_columns += [

    "AE_Cluster",

    "AE_Error_Type",

    "Direction_Pattern",

    "Extreme_Pattern",

    "Current",
    "Voltage",
    "Weld_Time",

    "Current_Error",
    "Voltage_Error",
    "Weld_Time_Error",

    "Reconstruction_Error"
]


detail = (
    candidate[
        detail_columns
    ]
    .sort_values(
        "Reconstruction_Error",
        ascending=False
    )
)


print("\n" + "=" * 90)
print("복원오차 상위 50개 실제 방향")
print("=" * 90)

print(
    detail
    .head(50)
    .round(4)
    .to_string(
        index=False
    )
)


# ============================================================
# 17. 방향 패턴 그래프
# ============================================================

top_patterns = (
    direction_count
    .head(15)
)


plt.figure(
    figsize=(14, 6)
)


plt.bar(
    top_patterns.index,
    top_patterns.values
)


plt.xlabel(
    "Actual Direction Pattern"
)

plt.ylabel(
    "Number of Rows"
)

plt.title(
    "Actual Process Direction of AE C1/C2 Candidates"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 18. 실제 Voltage 분포
# ============================================================

plt.figure(
    figsize=(12, 6)
)


for cluster in [
    0,
    1,
    2
]:

    temp = data[
        data["AE_Cluster"]
        == cluster
    ]


    plt.hist(
        temp["Voltage"],
        bins=40,
        alpha=0.4,
        label=f"C{cluster}"
    )


plt.axvline(
    baseline.loc[
        "Voltage",
        "median"
    ],
    linestyle="--",
    label="C0 Median"
)


plt.xlabel(
    "Voltage"
)

plt.ylabel(
    "Count"
)

plt.title(
    "Voltage Distribution by AE Cluster"
)

plt.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 19. 실제 Current 분포
# ============================================================

plt.figure(
    figsize=(12, 6)
)


for cluster in [
    0,
    1,
    2
]:

    temp = data[
        data["AE_Cluster"]
        == cluster
    ]


    plt.hist(
        temp["Current"],
        bins=40,
        alpha=0.4,
        label=f"C{cluster}"
    )


plt.axvline(
    baseline.loc[
        "Current",
        "median"
    ],
    linestyle="--",
    label="C0 Median"
)


plt.xlabel(
    "Current"
)

plt.ylabel(
    "Count"
)

plt.title(
    "Current Distribution by AE Cluster"
)

plt.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 20. 저장
# ============================================================

baseline.to_csv(
    "AE_C0_baseline.csv",
    encoding="utf-8-sig"
)


direction_summary.to_csv(
    "AE_actual_direction_summary.csv",
    encoding="utf-8-sig"
)


cluster_direction.to_csv(
    "AE_cluster_direction_pattern.csv",
    encoding="utf-8-sig"
)


detail.to_csv(
    "AE_candidate_actual_direction_detail.csv",
    index=False,
    encoding="utf-8-sig"
)


extreme_count.to_csv(
    "AE_extreme_pattern_count.csv",
    encoding="utf-8-sig"
)


# ============================================================
# 21. 최종 요약
# ============================================================

print("\n" + "=" * 90)
print("6단계 최종 결과")
print("=" * 90)


print(
    "전체 C1/C2 후보:",
    len(candidate)
)


print(
    "\n[Voltage 관련 후보]"
)

print(
    "전체:",
    len(voltage_candidate)
)

print(
    voltage_candidate[
        "Voltage_Direction"
    ]
    .value_counts()
)


print(
    "\n[Current 관련 후보]"
)

print(
    "전체:",
    len(current_candidate)
)

print(
    current_candidate[
        "Current_Direction"
    ]
    .value_counts()
)


print(
    "\n[Weld Time 관련 후보]"
)

print(
    "전체:",
    len(time_candidate)
)

print(
    time_candidate[
        "Weld_Time_Direction"
    ]
    .value_counts()
)


print(
    "\n[가장 많은 실제 방향 패턴 TOP 10]"
)

print(
    direction_summary
    .head(10)
    .round(2)
)


print(
    "\n※ Low/High는 C0 중앙값을 기준으로 한 상대적인 방향입니다."
)

print(
    "※ Extreme은 C0의 5~95% 범위를 벗어났다는 뜻이며,"
)

print(
    "   실제 공정 불량이라는 의미는 아닙니다."
)
```

```python
# ============================================================
# 7단계
# Force 공정조건별 AE 패턴 비교
#
# 목적
# ------------------------------------------------------------
# 현재 발견한 주요 AE 패턴이
#
#   Force 2.xxx
#       VS
#   Force 7~8
#
# 어느 공정조건에서 많이 나타나는지 확인
#
# 주요 패턴:
# 1. Voltage_High
# 2. Current_High + Voltage_High
# 3. Voltage_Low
# 4. Current_High
# 5. Current_High + Voltage_Low
#
# 주의:
# Force 그룹은 공정조건 구분용.
# AE 후보 = 실제 불량이라고 확정하지 않음.
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# 1. 6단계 candidate 확인
# ============================================================

required_cols = [
    "AE_Cluster",
    "Force",
    "Current",
    "Voltage",
    "Weld_Time",
    "Direction_Pattern"
]

missing = [
    col for col in required_cols
    if col not in candidate.columns
]

if missing:
    raise ValueError(
        f"candidate에 필요한 컬럼이 없습니다: {missing}\n"
        "6단계 코드를 먼저 실행하세요."
    )


# ============================================================
# 2. Force 그룹 생성
#
# 네가 지금까지 관찰한 두 공정조건:
#
# 2.xxx
# 7~8
#
# 그 외 값은 Other로 남겨서
# 억지로 두 그룹에 넣지 않음.
# ============================================================

candidate["Force_Group"] = np.select(

    [
        (candidate["Force"] >= 2.0)
        & (candidate["Force"] < 3.0),

        (candidate["Force"] >= 7.0)
        & (candidate["Force"] < 9.0)
    ],

    [
        "Force_2.xxx",
        "Force_7_8"
    ],

    default="Other"
)


# ============================================================
# 3. Force 그룹 전체 개수
# ============================================================

force_count = (
    candidate["Force_Group"]
    .value_counts()
)


force_rate = (
    candidate["Force_Group"]
    .value_counts(normalize=True)
    * 100
)


force_summary = pd.DataFrame({
    "Count": force_count,
    "Rate_%": force_rate
})


print("=" * 90)
print("C1/C2 후보의 Force 그룹")
print("=" * 90)

print(
    force_summary.round(2)
)


# ============================================================
# 4. 현재 주요 5개 AE 패턴
# ============================================================

main_patterns = [

    "Voltage_High",

    "Current_High + Voltage_High",

    "Voltage_Low",

    "Current_High",

    "Current_High + Voltage_Low"
]


main = candidate[
    candidate["Direction_Pattern"]
    .isin(main_patterns)
].copy()


print("\n" + "=" * 90)
print("주요 5개 AE 패턴")
print("=" * 90)

print(
    main["Direction_Pattern"]
    .value_counts()
)


print(
    "\n주요 5개 패턴 합계:",
    len(main)
)


# ============================================================
# 5. 핵심 결과
#
# AE 패턴 × Force 그룹 개수
# ============================================================

pattern_force = pd.crosstab(

    main["Direction_Pattern"],

    main["Force_Group"]
)


# 없는 그룹도 표시
for col in [
    "Force_2.xxx",
    "Force_7_8",
    "Other"
]:

    if col not in pattern_force.columns:
        pattern_force[col] = 0


pattern_force = pattern_force[
    [
        "Force_2.xxx",
        "Force_7_8",
        "Other"
    ]
]


pattern_force["Total"] = (
    pattern_force.sum(axis=1)
)


# 우리가 원하는 순서로 정렬
pattern_force = (
    pattern_force
    .reindex(main_patterns)
    .fillna(0)
    .astype(int)
)


print("\n" + "=" * 90)
print("주요 AE 패턴 × Force 그룹 [개수]")
print("=" * 90)

print(pattern_force)


# ============================================================
# 6. 각 AE 패턴 내부에서 Force 비율
#
# 예:
#
# Voltage_High 400개 중
#
# Force 2.xxx = 20%
# Force 7~8   = 70%
# Other       = 10%
#
# 이런 식으로 해석
# ============================================================

pattern_force_rate = (

    pattern_force[
        [
            "Force_2.xxx",
            "Force_7_8",
            "Other"
        ]
    ]

    .div(
        pattern_force["Total"],
        axis=0
    )

    * 100
)


pattern_force_rate["Total"] = 100


print("\n" + "=" * 90)
print("각 AE 패턴 내부 Force 구성 비율 (%)")
print("=" * 90)

print(
    pattern_force_rate
    .round(2)
)


# ============================================================
# 7. 반대로 Force 그룹 내부에서
# 어떤 AE 패턴이 많은지 확인
#
# 예:
#
# Force 7~8 후보들 중
# Voltage High가 몇 %인지
# ============================================================

force_pattern_rate = pd.crosstab(

    main["Force_Group"],

    main["Direction_Pattern"],

    normalize="index"

) * 100


force_pattern_rate = (
    force_pattern_rate
    .reindex(
        columns=main_patterns,
        fill_value=0
    )
)


print("\n" + "=" * 90)
print("Force 그룹 내부의 AE 패턴 비율 (%)")
print("=" * 90)

print(
    force_pattern_rate
    .round(2)
)


# ============================================================
# 8. C1 / C2까지 같이 비교
#
# Force × AE Cluster
# ============================================================

cluster_force = pd.crosstab(

    candidate["Force_Group"],

    candidate["AE_Cluster"]
)


print("\n" + "=" * 90)
print("Force 그룹 × AE Cluster")
print("=" * 90)

print(cluster_force)


# ============================================================
# 9. 주요 패턴 × Force × C1/C2
# ============================================================

pattern_force_cluster = pd.crosstab(

    [
        main["Direction_Pattern"],
        main["Force_Group"]
    ],

    main["AE_Cluster"]
)


print("\n" + "=" * 90)
print("AE 패턴 × Force × C1/C2")
print("=" * 90)

print(pattern_force_cluster)


# ============================================================
# 10. 실제 공정값 평균
#
# Force 그룹별
# Current / Voltage / Weld Time 실제 값 비교
# ============================================================

force_actual = (

    candidate

    .groupby("Force_Group")

    [
        [
            "Force",
            "Current",
            "Voltage",
            "Weld_Time"
        ]
    ]

    .agg(
        [
            "count",
            "mean",
            "median",
            "std",
            "min",
            "max"
        ]
    )
)


print("\n" + "=" * 90)
print("Force 그룹별 실제 공정값")
print("=" * 90)

print(
    force_actual.round(4)
)


# ============================================================
# 11. Force 그룹별 패턴 상세 평균
# ============================================================

pattern_actual = (

    main

    .groupby(
        [
            "Direction_Pattern",
            "Force_Group"
        ]
    )

    [
        [
            "Force",
            "Current",
            "Voltage",
            "Weld_Time"
        ]
    ]

    .agg(
        [
            "count",
            "mean",
            "median"
        ]
    )
)


print("\n" + "=" * 90)
print("AE 패턴 × Force별 실제 공정값")
print("=" * 90)

print(
    pattern_actual.round(4)
)


# ============================================================
# 12. 날짜 정보가 있으면
# 날짜 × Force 그룹도 같이 확인
# ============================================================

if "Date" in candidate.columns:

    date_force = pd.crosstab(

        candidate["Date"],

        candidate["Force_Group"]
    )


    print("\n" + "=" * 90)
    print("날짜별 Force 그룹의 AE 후보")
    print("=" * 90)

    print(date_force)


# ============================================================
# 13. 주요 패턴 × Force 그래프
# ============================================================

plot_data = (

    pattern_force[
        [
            "Force_2.xxx",
            "Force_7_8",
            "Other"
        ]
    ]
)


plot_data.plot(

    kind="bar",

    figsize=(14, 6)
)


plt.title(
    "Main AE Patterns by Force Condition"
)

plt.xlabel(
    "AE Pattern"
)

plt.ylabel(
    "Number of Rows"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 14. 비율 그래프
# ============================================================

pattern_force_rate[
    [
        "Force_2.xxx",
        "Force_7_8",
        "Other"
    ]
].plot(

    kind="bar",

    stacked=True,

    figsize=(14, 6)
)


plt.title(
    "Force Composition within Each AE Pattern"
)

plt.xlabel(
    "AE Pattern"
)

plt.ylabel(
    "Percentage (%)"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 15. 상세 데이터
# ============================================================

detail_cols = []


for col in [
    "Original_Row",
    "Date",
    "Working_Time"
]:

    if col in candidate.columns:
        detail_cols.append(col)


detail_cols += [

    "AE_Cluster",

    "Force_Group",

    "Direction_Pattern",

    "Force",
    "Current",
    "Voltage",
    "Weld_Time",

    "Reconstruction_Error"
]


force_detail = (

    candidate[
        detail_cols
    ]

    .sort_values(
        [
            "Force_Group",
            "Direction_Pattern"
        ]
    )
)


# ============================================================
# 16. 저장
# ============================================================

force_summary.to_csv(
    "AE_force_condition_summary.csv",
    encoding="utf-8-sig"
)


pattern_force.to_csv(
    "AE_pattern_by_force_count.csv",
    encoding="utf-8-sig"
)


pattern_force_rate.to_csv(
    "AE_pattern_by_force_rate.csv",
    encoding="utf-8-sig"
)


force_pattern_rate.to_csv(
    "AE_force_pattern_rate.csv",
    encoding="utf-8-sig"
)


force_detail.to_csv(
    "AE_force_pattern_detail.csv",
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 17. 최종 결과
# ============================================================

print("\n" + "=" * 90)
print("7단계 최종 결과")
print("=" * 90)


print("\n[전체 C1/C2의 Force 그룹]")

print(
    force_summary.round(2)
)


print("\n[주요 5개 패턴 × Force 개수]")

print(
    pattern_force
)


print("\n[각 패턴의 Force 구성 비율 %]")

print(
    pattern_force_rate.round(2)
)


print(
    "\n※ Force 2.xxx와 7~8은 "
    "현재 관찰된 공정조건을 비교하기 위한 그룹입니다."
)

print(
    "※ 특정 Force 그룹에 AE 후보가 많다고 해서 "
    "그 Force 조건이 실제 불량 원인이라는 의미는 아닙니다."
)
```

```python
# ============================================================
# 날짜별 AE 주요 패턴 + Force 조건 분석
# 기존 candidate 데이터만 사용
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ------------------------------------------------------------
# 1. candidate 복사
# ------------------------------------------------------------

x = candidate.copy()


# ------------------------------------------------------------
# 2. 날짜 확인
# ------------------------------------------------------------

if "Date" not in x.columns:
    raise ValueError("candidate에 Date 컬럼이 없습니다.")

x["Date"] = pd.to_datetime(
    x["Date"],
    errors="coerce"
).dt.strftime("%m/%d")


# ------------------------------------------------------------
# 3. Force 컬럼 찾기
# ------------------------------------------------------------

if "Force" not in x.columns:
    raise ValueError("candidate에 Force 컬럼이 없습니다.")


# ------------------------------------------------------------
# 4. Force 조건 분류
#
# 2.xxx
# 7~8
# 나머지 Other
# ------------------------------------------------------------

x["Force_Type"] = np.select(
    [
        (x["Force"] >= 2) & (x["Force"] < 3),
        (x["Force"] >= 7) & (x["Force"] < 9)
    ],
    [
        "Force_2.xxx",
        "Force_7_8"
    ],
    default="Other"
)


# ------------------------------------------------------------
# 5. 주요 AE 패턴
# ------------------------------------------------------------

patterns = [
    "Voltage_High",
    "Current_High + Voltage_High",
    "Voltage_Low",
    "Current_High",
    "Current_High + Voltage_Low"
]


# ------------------------------------------------------------
# 6. 날짜별 AE 패턴 개수
# ------------------------------------------------------------

pattern_data = x[
    x["Direction_Pattern"].isin(patterns)
].copy()


daily_pattern = pd.crosstab(
    pattern_data["Date"],
    pattern_data["Direction_Pattern"]
)


# 패턴이 없는 경우에도 컬럼 생성
for p in patterns:
    if p not in daily_pattern.columns:
        daily_pattern[p] = 0


daily_pattern = daily_pattern[patterns]


# ------------------------------------------------------------
# 7. 날짜별 전체 C1/C2 개수
# ------------------------------------------------------------

daily_total = (
    x.groupby("Date")
    .size()
    .rename("AE_C1_C2_Total")
)


# ------------------------------------------------------------
# 8. 날짜별 Force 조건 개수
# ------------------------------------------------------------

daily_force = pd.crosstab(
    x["Date"],
    x["Force_Type"]
)


for f in [
    "Force_2.xxx",
    "Force_7_8",
    "Other"
]:
    if f not in daily_force.columns:
        daily_force[f] = 0


daily_force = daily_force[
    [
        "Force_2.xxx",
        "Force_7_8",
        "Other"
    ]
]


# ------------------------------------------------------------
# 9. 전부 합치기
# ------------------------------------------------------------

result = (
    daily_pattern
    .join(daily_total, how="outer")
    .join(daily_force, how="outer")
    .fillna(0)
)


# 정수형
count_columns = (
    patterns
    + [
        "AE_C1_C2_Total",
        "Force_2.xxx",
        "Force_7_8",
        "Other"
    ]
)

result[count_columns] = (
    result[count_columns]
    .astype(int)
)


# ------------------------------------------------------------
# 10. 날짜 정렬
# ------------------------------------------------------------

result = result.sort_index()


# ------------------------------------------------------------
# 11. 결과 출력
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("날짜별 AE 패턴 + Force 조건")
print("=" * 110)

print(result.to_string())


# ------------------------------------------------------------
# 12. 보기 쉽게 날짜별 상세 출력
# ------------------------------------------------------------

for date, row in result.iterrows():

    print("\n" + "=" * 70)
    print("날짜:", date)
    print("=" * 70)

    print("\n[AE 주요 패턴]")

    print(
        "Voltage High:",
        row["Voltage_High"]
    )

    print(
        "Current High + Voltage High:",
        row["Current_High + Voltage_High"]
    )

    print(
        "Voltage Low:",
        row["Voltage_Low"]
    )

    print(
        "Current High:",
        row["Current_High"]
    )

    print(
        "Current High + Voltage Low:",
        row["Current_High + Voltage_Low"]
    )


    print("\n[전체 AE 후보]")

    print(
        "C1+C2:",
        row["AE_C1_C2_Total"]
    )


    print("\n[Force]")

    print(
        "Force 2.xxx:",
        row["Force_2.xxx"]
    )

    print(
        "Force 7~8:",
        row["Force_7_8"]
    )

    print(
        "Other:",
        row["Other"]
    )


# ------------------------------------------------------------
# 13. 날짜별 AE 패턴 그래프
# ------------------------------------------------------------

result[
    patterns
].plot(
    kind="bar",
    figsize=(15, 7)
)

plt.title(
    "Daily AE Pattern Count"
)

plt.xlabel(
    "Date"
)

plt.ylabel(
    "Count"
)

plt.xticks(
    rotation=45
)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 14. 날짜별 Force 그래프
# ------------------------------------------------------------

result[
    [
        "Force_2.xxx",
        "Force_7_8",
        "Other"
    ]
].plot(
    kind="bar",
    figsize=(15, 7)
)

plt.title(
    "Daily Force Condition"
)

plt.xlabel(
    "Date"
)

plt.ylabel(
    "Count"
)

plt.xticks(
    rotation=45
)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 15. 저장
# ------------------------------------------------------------

result.to_csv(
    "daily_AE_pattern_force.csv",
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 16. 전체 합계
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("전체 합계")
print("=" * 110)

print("\n[AE 주요 패턴]")

for p in patterns:
    print(
        p,
        ":",
        result[p].sum()
    )


print("\n[Force]")

print(
    "Force 2.xxx:",
    result["Force_2.xxx"].sum()
)

print(
    "Force 7~8:",
    result["Force_7_8"].sum()
)

print(
    "Other:",
    result["Other"].sum()
)


print("\n분석 완료")
```

```python
# ============================================================
# AE 주요 패턴 ↔ 실제 불량 3종 날짜별 비교
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. 실제 불량 데이터 직접 입력
# ============================================================

actual = pd.DataFrame({
    "Date": [
        "03/24",
        "03/25",
        "03/26",
        "03/30",
        "03/31",
        "04/02",
        "04/03",
        "04/07"
    ],

    "Pitting": [
        2, 3, 3, 1, 2, 1, 2, 0
    ],

    "Lack_Weld": [
        1, 2, 2, 2, 3, 1, 1, 1
    ],

    "Crack": [
        1, 3, 2, 1, None, 2, 1, 2
    ]
})


# ============================================================
# 2. candidate 복사
# ============================================================

x = candidate.copy()


# ============================================================
# 3. Date 형식 강제로 MM/DD로 통일
# ============================================================

def make_date(v):

    s = str(v)

    # 이미 03/25 형태
    if "/" in s and len(s) <= 5:
        return s

    try:
        return pd.to_datetime(v).strftime("%m/%d")
    except:
        return s


x["Date"] = x["Date"].apply(make_date)


# ============================================================
# 4. 사용할 AE 주요 패턴
# ============================================================

patterns = [
    "Voltage_High",
    "Current_High + Voltage_High",
    "Voltage_Low",
    "Current_High",
    "Current_High + Voltage_Low"
]


# ============================================================
# 5. 주요 패턴만 선택
# ============================================================

main = x[
    x["Direction_Pattern"].isin(patterns)
].copy()


# ============================================================
# 6. 날짜별 AE 패턴 개수
# ============================================================

daily_ae = pd.crosstab(
    main["Date"],
    main["Direction_Pattern"]
)


# 없는 컬럼 자동 생성
for p in patterns:

    if p not in daily_ae.columns:
        daily_ae[p] = 0


daily_ae = daily_ae[patterns]


# ============================================================
# 7. 날짜별 전체 C1+C2 후보
# ============================================================

total = (
    x.groupby("Date")
    .size()
    .rename("AE_Total")
)


daily_ae = daily_ae.join(
    total,
    how="outer"
).fillna(0)


# ============================================================
# 8. 실제 불량표 index 설정
# ============================================================

actual = actual.set_index("Date")


# ============================================================
# 9. AE + 실제 불량 결합
#
# inner 사용:
# AE와 실제 불량 날짜가 둘 다 존재하는 날짜만 사용
# ============================================================

compare = daily_ae.join(
    actual,
    how="inner"
)


# ============================================================
# 10. 결과 출력
# ============================================================

print("\n" + "=" * 100)
print("AE 패턴 ↔ 실제 불량 비교")
print("=" * 100)

print(compare.to_string())


# ============================================================
# 11. 실제 불량 총개수
# ============================================================

compare["Total_Defect"] = (
    compare["Pitting"]
    + compare["Lack_Weld"]
    + compare["Crack"].fillna(0)
)


print("\n" + "=" * 100)
print("날짜별 실제 불량 총개수")
print("=" * 100)

print(
    compare[
        [
            "Pitting",
            "Lack_Weld",
            "Crack",
            "Total_Defect"
        ]
    ]
)


# ============================================================
# 12. AE 패턴 ↔ 불량 상관
# ============================================================

defects = [
    "Pitting",
    "Lack_Weld",
    "Crack"
]


corr = pd.DataFrame(
    index=patterns,
    columns=defects
)


for p in patterns:

    for d in defects:

        temp = compare[
            [p, d]
        ].dropna()

        if (
            len(temp) >= 2
            and temp[p].nunique() > 1
            and temp[d].nunique() > 1
        ):

            corr.loc[p, d] = (
                temp[p]
                .corr(
                    temp[d],
                    method="spearman"
                )
            )

        else:

            corr.loc[p, d] = None


print("\n" + "=" * 100)
print("AE 패턴 ↔ 실제 불량 Spearman")
print("=" * 100)

print(corr)


# ============================================================
# 13. 날짜별 AE 패턴 출력
# ============================================================

print("\n" + "=" * 100)
print("날짜별 상세 비교")
print("=" * 100)


for date in compare.index:

    print("\n------------------------------")
    print("날짜:", date)
    print("------------------------------")

    print(
        "Voltage High:",
        int(compare.loc[date, "Voltage_High"])
    )

    print(
        "Current High + Voltage High:",
        int(compare.loc[
            date,
            "Current_High + Voltage_High"
        ])
    )

    print(
        "Voltage Low:",
        int(compare.loc[date, "Voltage_Low"])
    )

    print(
        "Current High:",
        int(compare.loc[date, "Current_High"])
    )

    print(
        "Current High + Voltage Low:",
        int(compare.loc[
            date,
            "Current_High + Voltage_Low"
        ])
    )


    print("\n실제 불량")

    print(
        "파임:",
        compare.loc[date, "Pitting"]
    )

    print(
        "용접부족:",
        compare.loc[date, "Lack_Weld"]
    )

    print(
        "크랙:",
        compare.loc[date, "Crack"]
    )


# ============================================================
# 14. 그래프 - 실제 불량
# ============================================================

compare[
    [
        "Pitting",
        "Lack_Weld",
        "Crack"
    ]
].plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Actual Defects by Date")
plt.xlabel("Date")
plt.ylabel("Defect Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ============================================================
# 15. 그래프 - AE 패턴
# ============================================================

compare[
    patterns
].plot(
    kind="bar",
    figsize=(14, 6)
)

plt.title("AE Patterns by Date")
plt.xlabel("Date")
plt.ylabel("AE Candidate Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ============================================================
# 16. 저장
# ============================================================

compare.to_csv(
    "AE_vs_actual_defect_final.csv",
    encoding="utf-8-sig"
)


print("\n" + "=" * 100)
print("완료")
print("=" * 100)

print("비교 날짜:", compare.index.tolist())

print(
    "\n※ 개별 AE 행과 개별 불량품을 연결한 것이 아니라"
    "\n   날짜별 AE 패턴과 날짜별 실제 불량을 비교한 결과입니다."
)
```
